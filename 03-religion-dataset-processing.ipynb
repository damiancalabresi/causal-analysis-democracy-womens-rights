# Pew Research Center - Religious Diversity Index

```python
import pandas as pd

rdi = pd.read_csv('processed-datasets/pew-research-center/religious-diversity-index-extended.csv')
print(f"Shape: {rdi.shape}")
rdi.head()
```

```python
democracy_wbl = pd.read_csv('processed-datasets/merged/democracy_wbl.csv')
democracy_iso3 = set(democracy_wbl['curr_iso3'].dropna().unique())

rdi_iso3 = set(rdi['iso3'].dropna().unique())

missing = sorted(rdi_iso3 - democracy_iso3)
present = sorted(rdi_iso3 & democracy_iso3)

print(f"RDI countries: {len(rdi_iso3)}")
print(f"Present in democracy_wbl: {len(present)}")
print(f"Missing from democracy_wbl ({len(missing)}): {missing}")
```

```python
merged = democracy_wbl.merge(
    rdi[['iso3', 'rdi', 'dominant_religion']],
    left_on='curr_iso3',
    right_on='iso3',
    how='left'
).drop(columns='iso3')

print(f"Shape: {merged.shape}")
print(f"rdi non-null: {merged['rdi'].notna().sum()} / {len(merged)}")
print(f"dominant_religion non-null: {merged['dominant_religion'].notna().sum()} / {len(merged)}")
merged[['curr_iso3', 'curr_country', 'year', 'rdi', 'dominant_religion']].head(10)
```

```python
merged.to_csv('processed-datasets/merged/democracy_wbl_rdi.csv', index=False)
print("Saved to processed-datasets/merged/democracy_wbl_rdi.csv")
```
